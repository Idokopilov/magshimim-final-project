```python
import pandas as pd
final_df = pd.read_csv('results_until_21000.csv')
final_df.head(10)
```

```python
import pandas as pd

# assuming your dataframe is df

df_long = final_df.melt(
    id_vars=["question", "index", "source"], 
    value_vars=["human_answers", "chatgpt_answers", "answer_gemini"],
    var_name="writer",   # this will now directly hold the column name
    value_name="answer"
)

# remove rows with missing answers
df_long = df_long.dropna(subset=["answer"])

df_long.head()
```

```python
# Shuffle all rows randomly
df_long = df_long.sample(frac=1, random_state=42).reset_index(drop=True)
```

```python
df_long.drop(columns=['index'], inplace=True)
```

```python
# reorder to [writer, question, answer, source, index]
df_long = df_long[["source", "question", "answer", "writer"]]
```

```python
df_long.to_csv('final_results_long_format.csv', index=False)
```
